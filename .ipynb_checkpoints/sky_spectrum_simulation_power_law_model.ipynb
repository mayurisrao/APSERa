### This notebook does the following
#### -- It generates a sky spectrum in Temperature vs frequency units over 2-4 GHz assuming a power law model of the sky
#### -- It assumes a sine^2"theta beam
#### -- It uses 3 input maps that are pre-prepared at spot frequencies and 5 deg resolution (3072 pixels in Healpix format
#### -- It uses an intensity vs frequency file for the CRR prediction 
#### -- Mimics for N spectra (vs time) the sky spectrum at a specific location and UTC -- with or without the CRR lines included
#### -- Calibration from intensity to temperature units assuming RJ correction, CMB temperature etc are all included
#### -- created spectra with and without CRR for a given location and LST
#### -- used Maximally smooth function to represent spectrally smooth foregrounds
#### -- used pysr to find the analytical equation of the residual of spectra and MS function
#### -- performed curvefit over the pysr equation and combined MS function+pysr equation
#### -- performed MCMC(emcee) on both combined equation and pysr equation 

#### Pre requisite packages and files to run this notebook
#### healpy, numpy, matplotlib, math, astropy, ephem, pandas, scipy, os
#### total_new_spec.txt, map_408_r4_5deg_nested_galactic_Kelvin.txt, map_1420_r4_5deg_nested_galactic_Kelvin.txt, map_23000_r4_5deg_nested_galactic_Kelvin.txt

### Define necessary constants and corrections (if any)

```python
## correction_150offset = 21.4; // Kelvin, Patra et al. 2015
## correction_150scaling = 1.05; // Patra et al. 2015
TCMB = 2.72548 # Kelvin
h = 6.62607157e-34
k = 1.3806488e-23
c = 2.99792458e+08
```

```python

def T_to_spectral_radiance_planck(freq_GHz,T):     #Freq in GHz
    return (2 * h * freq_GHz**3 * 1e+27 / c**2) / (np.exp((h * freq_GHz* 1e+9) / (k * T)) - 1)
def T_to_spectral_radiance_RJ(freq_GHz,T):     #Freq in GHz
    return (T * 2 * k * freq_GHz**2 * 1e+18 / c**2) 
def spectral_radiance_to_T_RJ(freq_GHz,I_nu):     #Freq in GHz
    return ((I_nu * c**2)/( 2 * k * freq_GHz**2 * 1e+18 )) 
```

```python
f_start = 2  # GHz              
f_stop = 4  #GHz
f_res = 0.001   # GHz 
N_SIDE = 16                   #Healpy parameter: resolution of the map is defined by the NSIDE parameter, which is generally a power of 2.
T_hot = 300
T_cold = 77                    #T_cold and T_hot in Kelvins (for atmospheric corrections to calculate noise added) purposefully taken to be a few thousands
latitude = 13.067439
longitude =80.237617            #Latitude and longitude in degrees
altitude = 0                     #Altitude in meters
alpha = 0.0065                   #Temp. lapse rate (deg C per meter); related to atmospheric corrections
PI = 3.14159265
Noise_int_time = 9.61e14
Trx = 30                        # Receiver temperature in Kelvin
T_atm = 1                      # Noise temp added by atmosphere 
freq = np.arange(f_start, f_stop, f_res)
freq_res_Hz = (freq[1]-freq[0])*1e9 # Hz
No_of_Channels = np.size(freq)
#create arrays 
CMB_Int = np.empty([No_of_Channels])
P_hot = np.empty([No_of_Channels])
P_cold = np.empty([No_of_Channels])
P_dif = np.empty([No_of_Channels])
NPIX = hp.nside2npix(N_SIDE)
print("Number of pixels for NSIDE ",N_SIDE," is ",NPIX)
print(
    "Approximate resolution at NSIDE {} is {:.2} deg".format(
        N_SIDE, hp.nside2resol(N_SIDE, arcmin=True) / 60
    )
)

#sky temperature in real and noise temperature stored in imaginary part
sigma = np.zeros([No_of_Channels])     #noise per channel
sigma_noCRR = np.zeros([No_of_Channels])
antenna_temp = np.zeros([No_of_Channels])
antenna_temp_noCRR = np.zeros([No_of_Channels])
noise_temp = np.zeros([No_of_Channels])
noise_temp_noCRR = np.zeros([No_of_Channels])
beam_weight = np.zeros([No_of_Channels,NPIX])

## Some defaults that can be changed later
n_spectra = 1                  #Number of spectra to be taken
delta_t_in_hours = 1/60   #Time step between consecutive spectra in hours

#Refraction correction formula
def refraction(alt_in_deg, altitude):
    alt_in_deg = np.asarray(alt_in_deg)
    pres = 1010.*math.pow((1-6.5/288000*altitude),5.255)
    if (altitude>11000):
        temp = 211.5
    else:
        temp = 283.0 - (alpha * altitude)
    R = 1.02/math.tan((alt_in_deg + (10.3/(alt_in_deg + 5.11)))*(PI/180))
    if (alt_in_deg == 90):
        R = R+ 0.0019279
    tpcor = (pres/1000) * (283/temp)
    R =  tpcor*(R/60)
    return alt_in_deg +R

# Obs = ephem.Observer()
# #print(Obs)
# t = Time(Time_in_UTC, scale='utc',location= EarthLocation(lat = latitude*u.deg, lon=longitude*u.deg, height=altitude*u.m))
# lst = t.sidereal_time('apparent')
# print(lst)

#alt == theta angle; az == phi angle; both in degrees
# alt == altitude angle (90 is zenith); az == azimuth angle; both in degrees
def beam_def(freq, alt, az, fwhm_deg=17.5, threshold=0.2):
    """
    Calculates the Gaussian beam response.
    Returns 0 if the pixel is below the horizon or below the power threshold.
    """
    if alt <= 0:
        return 0
    else:
        theta = 90.0 - alt
        sigma = fwhm_deg / 2.35482
        bw = np.exp(-(theta**2) / (2 * (sigma**2)))
        
        # New Truncation Logic:
        # If the gain is less than 'threshold' (e.g., 0.4), return 0.
        if bw < threshold:
            return 0
        return bw
```

### Generate the sky model -- option to save the model to a file in the file_output_path.
#### Fit the 3 input maps with a power law at each pixel, normalized to the first frequency. Obtain the spectral index
#### The power law is $T(\nu) = T_{\text{norm}} \left(\frac{\nu}{\nu_{\text{norm}}}\right)^{-\alpha}$.
#### Fit in log log mode $$ \log(T(\nu)) = \log(T_{\text{norm}}) - \alpha \, \log\!\left(\frac{\nu}{\nu_{\text{norm}}}\right)$$
#### Improvement needed - save the chisquare / goodness of fit as a second column in the output file

```python
power_law_model = os.path.join(file_output_path,"fitted_power_law_indices.txt")
# Example frequency array
input_freq = np.array([0.408, 1.420, 23.0])   # GHz
# Defines the output file path for the fitted parameters
alpha_output_file = os.path.join(file_output_path, "fitted_power_law_indices.txt")
chisq_output_file = os.path.join(file_output_path, "fitted_chi_square_values.txt")


# Example data arrays (replace with your real T1, T2, T3)
T1 = sky_408
T2 = sky_1420
T3 = sky_23000

# Stack temperature arrays -> shape (3072, 3)
T_all = np.vstack([T1, T2, T3]).T  # (rows=index, cols=freq)
errors= T_all*0.05  #this error should be replaced from original error from graph
errors[errors==0]=1e-10  #to avoid dividing 0 

alphas = []
chi_squares = []
for i, row in enumerate(T_all):
    # y values for this index
    y_vals = row  
    # Reference normalization (first freq, first T1 value for this index)
    x_norm = input_freq[0]
    y_norm = row[0]

    # log-log fit: log(y/y_norm) = -alpha * log(x/x_norm)
    logx = np.log(input_freq / x_norm)
    logy = np.log(y_vals / y_norm)

    # Linear fit slope = -alpha
    slope, _ = np.polyfit(logx, logy, 1)
    alpha = -slope
    alphas.append(alpha)
    T_expected = y_norm * (input_freq / x_norm) ** (-alpha) #to Calculate the expected temperatures predicted by the model using the fitted alpha
    T_observed = y_vals
    pixel_errors = errors[i] # to calculate the uncertainities in observed temperature
    chi2 = np.sum(((T_observed - T_expected) / pixel_errors)**2) #to calculate the chi squared values
    chi_squares.append(chi2)
alphas = np.array(alphas)
chi_squares = np.array(chi_squares)
#output_data = np.vstack([alphas, chi_squares]).T #(to save alpha, chi squared values in the same file)

# Save as a text file (one column of alphas)
#np.savetxt(output_model_file, output_data, fmt="%.4f", header="Alpha\tChi_Square", delimiter="\t") #(to save in same file)
print("Fitting done. Saved alpha and chi-square values to fitted_power_law_indices_and_chisq.txt in output path")
np.savetxt(alpha_output_file, alphas, fmt="%.4f", header="Alpha") #to save in saperate file
np.savetxt(chisq_output_file, chi_squares, fmt="%.4f", header="Chi_Square") # to save in saparate file
```

```python
pixel_indices = np.arange(0,3072)
coords = hp.pixelfunc.pix2ang(16, pixel_indices , nest = True, lonlat = True)          #pix2ang = gives l and b coordinates for each pixel; NSIDE=16 
gc = SkyCoord(l = coords[0]*u.degree, b= coords[1]*u.degree, frame = 'galactic') 
print(np.shape(coords))
print(np.shape(gc))
```

```python
CMB_Int = T_to_spectral_radiance_planck(freq, TCMB)
input_freq = np.array([0.408, 1.420, 23.0]) # GHz
f_norm = input_freq[0]
T_norm = sky_408
power_law_model = os.path.join(file_output_path, "fitted_power_law_indices.txt")
alphas = np.loadtxt(power_law_model)
current_utc_time = spectrum_UTC_today[0] 
obs_loc = EarthLocation(lat=13.601*u.deg, lon=77.428*u.deg, height=690.0*u.m)

ant_T = np.zeros([No_of_Channels])
sumwt = np.zeros([No_of_Channels])
ant_T_noCRR = np.zeros([No_of_Channels])
sumwt_noCRR = np.zeros([No_of_Channels])

equi = gc.transform_to(AltAz(obstime=current_utc_time, location=obs_loc))
alt = equi.alt.value
az = equi.az.value

for pix in range(NPIX):
    refined_alt = refraction(alt[pix], obs_loc.height.value)
    alpha = alphas[pix]
    
    for ch in range(No_of_Channels):
        cwt = beam_def(freq[ch], refined_alt, az[pix])
        
        if cwt > 0:
            # 1. Sky Model (Power Law)
            sky_T = T_norm[pix] * ((freq[ch] / f_norm) ** (-alpha))
            foreground_rad = T_to_spectral_radiance_RJ(freq[ch], sky_T)
            
        
           # 2a. WITH CRR
            rad_total = foreground_rad + CMB_Int[ch] + CRR_Int[ch]
            pwr_total = (rad_total * (c*c) * freq_res_Hz) / (1e+18 * freq[ch]**2)
            T_final = (pwr_total / P_dif[ch]) * (T_hot - T_cold)
            ant_T[ch] += cwt * T_final
            sumwt[ch] += cwt

            # 2b. WITHOUT CRR — needs its own pwr_total / T_final
            rad_total_noCRR = foreground_rad + CMB_Int[ch]
            pwr_total_noCRR = (rad_total_noCRR * (c*c) * freq_res_Hz) / (1e+18 * freq[ch]**2)
            T_final_noCRR = (pwr_total_noCRR / P_dif[ch]) * (T_hot - T_cold)
            ant_T_noCRR[ch] += cwt * T_final_noCRR
            sumwt_noCRR[ch] += cwt

sumwt[sumwt == 0] = 1.0 
ant_T /= sumwt
sumwt_noCRR[sumwt_noCRR == 0] = 1.0
ant_T_noCRR /= sumwt_noCRR 
sig = (ant_T + Trx + T_atm) / math.sqrt(freq_res_Hz * Noise_int_time)
sig_noCRR = (ant_T_noCRR + Trx + T_atm) / math.sqrt(freq_res_Hz * Noise_int_time)

noise_random = np.random.normal(0, 1, No_of_Channels)
spectrum_with_CRR = ant_T + (noise_random * sig)
spectrum_no_CRR   = ant_T_noCRR + (noise_random * sig_noCRR)  
fig, ax = plt.subplots(figsize=(8, 7))
ax.plot(freq, spectrum_with_CRR, color='tab:blue', lw=1.2)
for spine in ax.spines.values():
    spine.set_linewidth(2.5)
    spine.set_color('black')
ax.set_xlabel('Frequency [GHz]', fontsize=12)
ax.set_ylabel('Temperature [K]', fontsize=12)
t_obs = Time(current_utc_time, scale='utc', location=obs_loc)
lst_val = t_obs.sidereal_time('apparent').to_value(u.hourangle)
ax.set_title(f"Site: Gauribidnur\nLST: {lst_val:.1f}h with", fontsize=14, pad=15)
ax.grid(True, linestyle='-', color='gray', alpha=0.7)
ax2 = ax.twinx()
ax2.set_ylim(ax.get_ylim())
ax2.set_ylabel('Temperature [K]', fontsize=12)
for spine in ax2.spines.values():
    spine.set_linewidth(2.5)
    spine.set_color('black')

ax.tick_params(axis='both', which='major', labelsize=11)
ax2.tick_params(axis='y', which='major', labelsize=11)

plt.show()
```

```python
# Fit smooth function in log-log space
#
# Compute derivatives and if any zero crossings are deemed present 
# in the range then blow up the chisq if so.
#

# *****IMPORTANT*****
# The values of X1 and X2 need to be changed to be the end limits 
# of the frequency range in GHz.
# *****IMPORTANT*****

import numpy as np
import matplotlib.pyplot as plt
import scipy.constants
import math as math
from math import exp, expm1, sqrt
from math import factorial as mf
from scipy.optimize import fmin

PI=scipy.constants.pi
HH=scipy.constants.h
KK=scipy.constants.k
HbK=HH/KK
HbK9=HbK*1.0e9


# X1 = 1.0  # Original
# X2 = 7.0  # Original
X1 = f_start # Sourced directly from simulation parameters
X2 = f_stop  # Sourced directly from simulation parameters


print (' ')
print ('CAUTION: this version is hard-coded for freq range: ',X1,' to ',X2,' GHz')

X1LOG = np.log10(X1)
X2LOG = np.log10(X2)
XDIF = X1LOG-X2LOG

np.set_printoptions(precision=20)

func11 = lambda p, x: (HbK9*x)/( np.exp( HbK9*x/(10.0**p[0]) )-1.0 ) + \
                    10.0**( (p[1]) + (p[2])*(np.log10(x)-X1LOG)**1 )

func22 = lambda p, x: (HbK9*x)/( np.exp( HbK9*x/(10.0**p[0]) )-1.0 ) +  \
                    10.0**( (p[1]) + (p[2])*(np.log10(x)-X1LOG)**1 + \
                    (p[3])*(np.log10(x)-X1LOG)**2 )

func33 = lambda p, x: (HbK9*x)/( np.exp( HbK9*x/(10.0**p[0]) )-1.0 ) + \
                    10.0**( (p[1]) + (p[2])*(np.log10(x)-X1LOG)**1 + \
                    (p[3])*(np.log10(x)-X1LOG)**2 + (p[4])*(np.log10(x)-X1LOG)**3 )

func44 = lambda p, x: (HbK9*x)/( np.exp( HbK9*x/(10.0**p[0]) )-1.0 ) + \
                    10.0**( (p[1]) + (p[2])*(np.log10(x)-X1LOG)**1 + \
                    (p[3])*(np.log10(x)-X1LOG)**2 + (p[4])*(np.log10(x)-X1LOG)**3 + \
                    (p[5])*(np.log10(x)-X1LOG)**4 )

chisq11 = lambda p, x, y: sqrt(((func11(p,x)-y)**2).sum()/float(len(x)))
chisq22 = lambda p, x, y: sqrt(((func22(p,x)-y)**2).sum()/float(len(x)))

def chisq33 (p, x, y):
    k2 = (mf(2)/mf(0))*p[3] + (mf(3)/mf(1))*p[4]*(np.log10(x)-X1LOG)
    for i in range (len(x)-1):
        if k2[i+1]*k2[i] < 0.0: 
            return 100.0+min(i,len(x)-i)
    return sqrt(((func33(p,x)-y)**2).sum()/float(len(x)))

def chisq44 (p, x, y):
    k2 = (mf(2)/mf(0))*p[3] + (mf(3)/mf(1))*p[4]*(np.log10(x)-X1LOG) + \
        (mf(4)/mf(2))*p[5]*(np.log10(x)-X1LOG)**2
    k3 = (mf(3)/mf(0))*p[4] + (mf(4)/mf(1))*p[5]*(np.log10(x)-X1LOG)
    for i in range (len(x)-1):
        if k2[i+1]*k2[i] < 0.0 or k3[i+1]*k3[i] < 0.0: 
            return 100.0+min(i,len(x)-i)
    return sqrt(((func44(p,x)-y)**2).sum()/float(len(x)))

# f = open('spec_to_get_template_6nov15.txt','r')
# line1 = f.readline()
# line1 = line1.strip()
# x1 = line1.split()
# x1 = np.asarray(x1, dtype=np.float64)
# x1log = np.log10(x1)
# irecord=0
# for line2 in f:
#     print (" processing record number ",irecord)
#     line2 = line2.strip()
#     y1 = line2.split()
#     y1 = np.asarray(y1, dtype=np.float64)
#     y1log = np.log10(y1)

x1 = freq
x1log = np.log10(x1)

y1 = spectrum_with_CRR
y2 = spectrum_no_CRR
y1log = np.log10(y1)

print ("\n Processing simulated spectrum...")

p00 = [np.log10(3.0),1.0,-2.5] 

p1 = fmin(chisq11, p00, args=(x1,y1),ftol=1.0e-20,maxiter=50000, maxfun=100000, disp=False) 
current_chisq = chisq11(p1,x1,y1)
print (" \nchisq1 = ",current_chisq)
print (10.0**p1[0], p1[1], p1[2])

p00 = [p1[0],p1[1],p1[2],0.0] 
p1 = fmin(chisq22, p00, args=(x1,y1),ftol=1.0e-20,maxiter=50000, maxfun=100000, disp=False) 
final_chisq = chisq22(p1,x1,y1)
print (" \nchisq2 = ",final_chisq)
print (10.0**p1[0], p1[1], p1[2], p1[3])

p00 = [p1[0],p1[1],p1[2],p1[3],0.0]
p1 = fmin(chisq33, p00, args=(x1,y1),ftol=1.0e-20,maxiter=50000, maxfun=100000, disp=False) 
final_chisq = chisq33(p1,x1,y1)
print (" \nchisq3 = ",final_chisq)
print (10.0**p1[0], p1[1], p1[2], p1[3], p1[4])

p00 = [p1[0],p1[1],p1[2],p1[3],p1[4],0.0]
p1 = fmin(chisq44, p00, args=(x1,y1),ftol=1.0e-20,maxiter=50000, maxfun=100000, disp=False) 
final_chisq = chisq44(p1,x1,y1)
print (" \nchisq4 = ",final_chisq)
print (10.0**p1[0], p1[1], p1[2], p1[3], p1[4], p1[5])
print (" \n")

yfit = func44(p1,x1)     # This fit is of foreground 
yres = y1 - yfit         # This residual is for H0
yres = np.asarray(yres)

plt.figure(figsize=(10, 7)) 
plt.grid(linewidth=1.5, linestyle='--', alpha=0.7) 
plt.plot(x1, yres*1.0e9, linestyle='-', lw=2, color='#E97451') 

ax = plt.gca()
for spine in ax.spines.values():
    spine.set_linewidth(2.5)
    spine.set_color('black')

plt.tick_params(axis='both', which='major', labelsize=14)
plt.tick_params(axis='both', which='minor', labelsize=14)
plt.xlabel('Frequency [GHz]', fontsize=14, fontweight='bold')
plt.ylabel('Recombination line signal [nK]', fontsize=14, fontweight='bold')
plt.title('Order 4 MSfit Polynomial Subtraction', fontsize=16, fontweight='bold')



plt.show()
```

# Curve fit for the pysr equation

```python
import numpy as np
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt

def dTb_pysr_model_8p(x0, p0, p1, p2, p3, p4, p5, p6, p7):
    exp_term = np.exp(p3 / np.log(x0))
    
    # Numerator term
    arg_num = p1 * (x0 + p2) * (x0 - exp_term) - p4
    numerator = -p0 * (np.cos(arg_num) + p5)
    
    # Denominator term
    arg_den = x0 * (-x0 + exp_term)
    denominator = p6 * x0 + np.cos(x0 * np.log(x0)) - p7 + (p6 * np.cos(arg_den)) / x0
    
    # Guard against division by zero/near-zero poles
    denom_clipped = np.where(np.abs(denominator) < 1e-12, np.sign(denominator) * 1e-12, denominator)
    return numerator / denom_clipped

p0_initial = [
    1.8430768e-8,       # p0: Amplitude scale
    0.98813858,         # p1: Numerator chirp frequency
    0.03189955,         # p2: Frequency shift
    1.5563995,          # p3: Exponent scale in exp(p3 / ln(x0))
    0.41928691,         # p4: Numerator phase offset
    0.07998569,         # p5: Numerator DC offset
    0.90427417,         # p6: Denominator scaling factor
    0.23749791          # p7: Denominator constant offset
]

lower_bounds = [
    0.0,     0.1,  -1.0,   0.1,  -np.pi, -1.0,   0.1,  -5.0
]
upper_bounds = [
    1.0e-5, 10.0,   1.0,  10.0,   np.pi,  1.0,  10.0,   5.0
]

freq_arr = np.asarray(freq)
yres_arr = np.asarray(yres)

# Harmonize array dimensions if there is a length mismatch
if freq_arr.shape != yres_arr.shape:
    if yres_arr.shape[0] == 2000 and freq_arr.shape[0] != 2000:
        freq_fit = np.arange(2.0, 4.0, 0.001)
        yres_fit = yres_arr
    elif freq_arr.shape[0] == 150 and yres_arr.shape[0] != 150:
        full_grid = np.arange(2.0, 4.0, 0.001)
        yres_fit = np.interp(freq_arr, full_grid, yres_arr)
        freq_fit = freq_arr
    else:
        raise ValueError(f"Shape mismatch: freq {freq_arr.shape} vs yres {yres_arr.shape}")
else:
    freq_fit = freq_arr
    yres_fit = yres_arr

popt_8p, pcov_8p = curve_fit(
    dTb_pysr_model_8p,
    freq_fit,
    yres_fit,
    p0=p0_initial,
    bounds=(lower_bounds, upper_bounds),
    method='trf',
    x_scale='jac',
    ftol=1e-15,
    xtol=1e-15,
    maxfev=150000
)

perr_8p = np.sqrt(np.diag(pcov_8p))

param_names = [
    'p0 (Amp Scale)',
    'p1 (Chirp Freq)',
    'p2 (Shift)',
    'p3 (Exp Scale)',
    'p4 (Phase Offset)',
    'p5 (DC Offset)',
    'p6 (Denom Scale)',
    'p7 (Denom Offset)'
]

print("="*65)
print(f"{'Parameter':<20} | {'Fitted Value':<18} | {'1-Sigma Error':<18}")
print("="*65)
for name, val, err in zip(param_names, popt_8p, perr_8p):
    print(f"{name:<20} | {val:<18.7e} | ± {err:<18.7e}")
print("="*65)

y_fit = dTb_pysr_model_8p(freq_fit, *popt_8p)
residuals = yres_fit - y_fit
sigma_res = np.std(residuals)

fig, (ax1, ax2) = plt.subplots(
    2, 1, figsize=(9, 6), sharex=True, 
    gridspec_kw={'height_ratios': [3, 1], 'hspace': 0.08}
)

ax1.plot(freq_fit, yres_fit, '.', color='#4A5568', markersize=2, alpha=0.5, label='Residual Data')
ax1.plot(freq_fit, y_fit, color='#E53E3E', lw=1.8, label='8-Parameter PySR Fit')
ax1.set_ylabel('Residual [K]', fontsize=12)
ax1.grid(True, linestyle='--', alpha=0.6)
ax1.legend(loc='upper right', fontsize=10)
ax1.set_title('8-Parameter PySR Fit Optimization', fontsize=13)

# Normalized pulls
pulls = residuals / sigma_res
ax2.plot(freq_fit, pulls, '.', color='#2D3748', markersize=2, alpha=0.6)
ax2.axhline(0, color='#E53E3E', ls='--', lw=1.0)
ax2.axhline(1, color='gray', ls=':', lw=0.8)
ax2.axhline(-1, color='gray', ls=':', lw=0.8)
ax2.set_ylabel(r'Pull [$\sigma$]', fontsize=11)
ax2.set_xlabel('Frequency [GHz]', fontsize=12)
ax2.set_ylim(-4, 4)
ax2.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

# MCMC for curve fit pysr residual


```python
import emcee
import corner
import numpy as np
import matplotlib.pyplot as plt

def dTb_pysr_model_8p_mcmc(x0, log10_p0, p1, p2, p3, p4, p5, p6, p7):
    p0 = 10.0**log10_p0
    exp_term = np.exp(p3 / np.log(x0))
    
    # Numerator
    arg_num = p1 * (x0 + p2) * (x0 - exp_term) - p4
    numerator = -p0 * (np.cos(arg_num) + p5)
    
    # Denominator
    arg_den = x0 * (-x0 + exp_term)
    denominator = p6 * x0 + np.cos(x0 * np.log(x0)) - p7 + (p6 * np.cos(arg_den)) / x0
    
    denom_clipped = np.where(np.abs(denominator) < 1e-12, np.sign(denominator) * 1e-12, denominator)# 8 parameter PySR model for MCMC sampling
    return numerator / denom_clipped

param_labels_mcmc = [
    r"$\log_{10}(p_0)$",
    r"$p_1$",
    r"$p_2$",
    r"$p_3$",
    r"$p_4$",
    r"$p_5$",
    r"$p_6$",
    r"$p_7$"
]
ndim = 8 # total number of parameters for MCMC sampling

lower_bounds_mcmc = np.array([-11.0,  0.1, -1.0,  0.1, -np.pi, -1.0,  0.1, -5.0])# lower bounds for MCMC sampling form curve_fit results
upper_bounds_mcmc = np.array([ -5.0, 10.0,  1.0, 10.0,  np.pi,  1.0, 10.0,  5.0])# upper bounds for MCMC sampling from curve_fit results

def log_prior(theta):
    if np.all((theta >= lower_bounds_mcmc) & (theta <= upper_bounds_mcmc)):
        return 0.0
    return -np.inf # prior is uniform within bounds and -inf outside

def log_likelihood(theta, x, y, yerr):
    model = dTb_pysr_model_8p_mcmc(x, *theta)
    if not np.all(np.isfinite(model)):
        return -np.inf
    inv_sigma2 = 1.0 / (yerr**2)
    return -0.5 * np.sum((y - model)**2 * inv_sigma2) # log-likelihood assuming Gaussian errors

def log_probability(theta, x, y, yerr):
    lp = log_prior(theta)
    if not np.isfinite(lp):
        return -np.inf
    ll = log_likelihood(theta, x, y, yerr)
    return lp + ll if np.isfinite(ll) else -np.inf # log-probability is sum of log-prior and log-likelihood, with checks for finiteness
theta_mode = np.array([
    np.log10(popt_8p[0]),
    popt_8p[1],
    popt_8p[2],
    popt_8p[3],
    popt_8p[4],
    popt_8p[5],
    popt_8p[6],
    popt_8p[7]
]) # Mode of the posterior distribution from curve_fit results, with log10 transformation for p0 because of its wide dynamic range. This serves as the starting point for MCMC sampling.

model_fit = dTb_pysr_model_8p_mcmc(freq_fit, *theta_mode)# Compute the model fit using the mode of the posterior distribution obtained from curve_fit. This serves as a reference for calculating residuals and estimating the noise floor in the data.
residuals = yres_fit - model_fit# Calculate the residuals by subtracting the model fit from the observed data. These residuals are used to estimate the noise characteristics of the data, which is crucial for setting up the MCMC sampling process.
sigma_noise = np.std(residuals) # Estimate the standard deviation of the residuals, which serves as an estimate of the noise floor in the data. This value is used to define the uncertainty (yerr) for each data point in the MCMC sampling process.
yerr = np.full_like(yres_fit, sigma_noise) #    Create an array of uncertainties (yerr) for each data point in the residuals, all set to the estimated noise floor (sigma_noise). This uniform uncertainty is used in the log-likelihood calculation during MCMC sampling.

print(f"Per-channel noise floor (RMS): {sigma_noise:.4e} K")

nwalkers = 32 # Number of walkers for MCMC sampling. A higher number of walkers can improve the exploration of the parameter space but increases computational cost.
max_steps = 15000 # Maximum number of steps for MCMC sampling. This sets an upper limit on the number of iterations the sampler will perform, allowing for convergence checks and preventing infinite loops in case of non-convergence.
check_interval = 200 # Interval at which to check for convergence during MCMC sampling. The sampler will evaluate the autocorrelation time and convergence criteria every 'check_interval' steps, allowing for early termination if convergence is achieved before reaching 'max_steps'.

np.random.seed(42) # Set a random seed for reproducibility of the MCMC sampling process. This ensures that the random initialization of walkers and any stochastic processes yield the same results across different runs.
pos = theta_mode + 1e-4 * np.abs(theta_mode) * np.random.randn(nwalkers, ndim) # Initialize the positions of the walkers for MCMC sampling. Each walker is initialized around the mode of the posterior distribution (theta_mode) with a small random perturbation (1e-4 times the absolute value of theta_mode) to ensure diversity in the starting positions while remaining close to the best-fit solution.

# Clamp walkers strictly within prior boundaries
for i in range(ndim):
    pos[:, i] = np.clip(pos[:, i], lower_bounds_mcmc[i] + 1e-5, upper_bounds_mcmc[i] - 1e-5) # Ensure that all walkers are initialized strictly within the defined prior boundaries. This prevents any walkers from starting outside the valid parameter space, which could lead to invalid log-probability evaluations and hinder convergence.

sampler = emcee.EnsembleSampler(
    nwalkers,
    ndim,
    log_probability,
    args=(freq_fit, yres_fit, yerr)
) #    Create an instance of the emcee EnsembleSampler, which will perform the MCMC sampling. The sampler is initialized with the number of walkers, number of dimensions (parameters), the log-probability function, and the data (frequency, residuals, and uncertainties) to be used in the sampling process.

print(f"\nSampling with {nwalkers} walkers on {ndim} parameters...")
old_tau = np.inf  # Initialize the previous autocorrelation time (old_tau) to infinity. This variable will be used to track the change in autocorrelation time between convergence checks, allowing for the assessment of whether the sampler has stabilized.
converged = False # Initialize the convergence flag to False. This flag will be set to True if the sampler meets the convergence criteria based on the autocorrelation time and its relative change, indicating that the MCMC sampling has sufficiently explored the parameter space.

for sample in sampler.sample(pos, iterations=max_steps, progress=True):
    if sampler.iteration % check_interval != 0 or sampler.iteration < 600:
        continue

    tau = sampler.get_autocorr_time(tol=0) # Compute the autocorrelation time (tau) of the sampler at the current iteration. The autocorrelation time provides an estimate of how many steps are needed for the samples to become effectively independent, which is crucial for assessing convergence and determining the effective sample size.
    converged = np.all(tau * 50 < sampler.iteration) # Check if the sampler has converged by verifying that the autocorrelation time multiplied by 50 is less than the current iteration count for all parameters. This condition ensures that the sampler has run for a sufficient number of iterations relative to the autocorrelation time, indicating that the samples are likely representative of the posterior distribution.
    converged &= np.all(np.abs(old_tau - tau) / np.where(tau > 0, tau, 1.0) < 0.01) # Further check for convergence by ensuring that the relative change in autocorrelation time between the current and previous checks is less than 1% for all parameters. This condition helps confirm that the sampler has stabilized and is not experiencing significant fluctuations in autocorrelation time, which could indicate ongoing exploration of the parameter space.
    old_tau = tau # Update the previous autocorrelation time (old_tau) to the current value of tau for use in the next convergence check. This allows for tracking changes in autocorrelation time across iterations, which is essential for assessing convergence stability.

    if converged:
        print(f"\nEnsemble converged at step {sampler.iteration}!") # Print a message indicating that the ensemble has converged.
        break

if not converged:
    print(f"\nReached max iterations ({max_steps}) before N > 50*tau convergence.")

tau = sampler.get_autocorr_time(quiet=True) # Compute the final autocorrelation time (tau) of the sampler after completing the MCMC sampling. This value provides an estimate of how many steps are needed for the samples to become effectively independent, which is important for determining the effective sample size and assessing the quality of the posterior samples.
valid_tau = np.all(np.isfinite(tau)) # Check if all values of the autocorrelation time (tau) are finite. This validation step ensures that the computed autocorrelation times are valid and can be used for further analysis, such as determining burn-in and thinning parameters for the MCMC samples.

burnin = int(2 * np.max(tau)) if valid_tau else 1000 # Determine the burn-in period for the MCMC samples based on the autocorrelation time (tau). The burn-in period is set to twice the maximum autocorrelation time if valid tau values are available; otherwise, a default value of 1000 steps is used. The burn-in period represents the initial portion of the MCMC chain that is discarded to allow the sampler to reach a stable state and avoid bias from initial conditions.
thin = int(0.5 * np.min(tau)) if valid_tau else 15 # Determine the thinning factor for the MCMC samples based on the autocorrelation time (tau). The thinning factor is set to half of the minimum autocorrelation time if valid tau values are available; otherwise, a default value of 15 steps is used. Thinning is applied to reduce autocorrelation in the samples by retaining only every 'thin' step, which helps ensure that the retained samples are more independent and representative of the posterior distribution.
thin = max(1, thin)

print("\n--- Diagnostics Summary ---")
print(f"Mean Acceptance Fraction : {np.mean(sampler.acceptance_fraction):.3f}")
for name, t in zip(param_labels_mcmc, tau):
    print(f"Parameter {name:16s} : tau = {t:6.2f} steps")
print(f"Burn-in Discarded        : {burnin} steps")
print(f"Thinning Factor          : Every {thin} steps")

# Walker trace chains
chain = sampler.get_chain() # Retrieve the full MCMC chain of samples from the sampler. The chain contains the sampled parameter values for each walker at each iteration, which can be used for diagnostics, visualization, and posterior analysis.
fig, axes = plt.subplots(ndim, 1, figsize=(10, 1.7 * ndim), sharex=True)
for i in range(ndim):
    axes[i].plot(chain[:, :, i], color="k", alpha=0.25, lw=0.6)
    axes[i].axhline(theta_mode[i], color="#D83A56", linestyle="--", lw=1.2)
    axes[i].axvline(burnin, color="blue", linestyle=":", lw=1.2)
    axes[i].set_ylabel(param_labels_mcmc[i], fontsize=11)
    axes[i].yaxis.set_label_coords(-0.08, 0.5)

axes[-1].set_xlabel("Iteration Step", fontsize=11)
plt.tight_layout()
plt.show()

flat_samples = sampler.get_chain(discard=burnin, thin=thin, flat=True)

# Convert log10(p0) to physical amplitude
samples_physical = np.copy(flat_samples)
samples_physical[:, 0] = 10.0**samples_physical[:, 0]

labels_physical = [r"$p_0$", r"$p_1$", r"$p_2$", r"$p_3$", r"$p_4$", r"$p_5$", r"$p_6$", r"$p_7$"]

print("\n--- Physical Parameter Constraints (Median ± 1-sigma) ---")
for i in range(ndim):
    percentiles = np.percentile(samples_physical[:, i], [16, 50, 84])
    q = np.diff(percentiles)
    print(f"{labels_physical[i]:12s} : {percentiles[1]:14.7e} (+{q[1]:.7e}, -{q[0]:.7e})")

fig = corner.corner(
    flat_samples,
    labels=param_labels_mcmc,
    truths=theta_mode,
    truth_color="#D83A56",
    quantiles=[0.16, 0.50, 0.84],
    show_titles=True,
    title_fmt=".3f",
    title_kwargs={"fontsize": 9},
    smooth=1.0,
    plot_datapoints=False,
    fill_contours=True,
    levels=(0.68, 0.95)
)
plt.show()
```

# Curve fit for 12 parameter equation

```python
import numpy as np
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt

def ms_pysr_combined_model(nu, a0, a1, a2, a3, a4, p0, p1, p2, p3, p4, p5, p6, p7): # Combined model function for fitting the residual spectrum with a maximally smooth polynomial and an 8-parameter PySR model.
    # --- Part 1: Maximally Smooth (MS) Order-4 Polynomial Baseline ---
    # Centered around the band midpoint (3.0 GHz) to orthogonalize polynomial coefficients
    x = nu - 3.0
    t_ms = a0 + a1 * x + a2 * x**2 + a3 * x**3 + a4 * x**4
    
    # --- Part 2: 8-Parameter PySR Non-Linear Model ---
    exp_term = np.exp(p3 / np.log(nu))
    
    # Numerator
    arg_num = p1 * (nu + p2) * (nu - exp_term) - p4
    numerator = -p0 * (np.cos(arg_num) + p5)
    
    # Denominator
    arg_den = nu * (-nu + exp_term)
    denominator = p6 * nu + np.cos(nu * np.log(nu)) - p7 + (p6 * np.cos(arg_den)) / nu
    denom_clipped = np.where(np.abs(denominator) < 1e-12, np.sign(denominator) * 1e-12, denominator)
    
    t_pysr = numerator / denom_clipped
    
    return t_ms + t_pysr

freq_arr = np.asarray(freq)
# Target is the total measured/simulated spectrum (e.g., ~2.68 to 2.93 K curve)
y_target = np.asarray(spectrum_with_CRR if 'spectrum_with_CRR' in locals() else yres)

# Guard against array dimension mismatches
if freq_arr.shape != y_target.shape:
    if y_target.shape[0] == 2000 and freq_arr.shape[0] != 2000: # to avoid interpolation of the 2000-point spectrum to the 150-point frequency grid, we fit on the full 2000-point grid
        freq_fit = np.linspace(2.0, 4.0, 2000)
    elif freq_arr.shape[0] == 150 and y_target.shape[0] != 150:
        full_grid = np.linspace(2.0, 4.0, 2000)
        y_target = np.interp(freq_arr, full_grid, y_target)
        freq_fit = freq_arr# to avoid interpolation of the 2000-point spectrum to the 150-point frequency grid, we fit on the full 2000-point grid
    else:
        freq_fit = np.linspace(2.0, 4.0, len(y_target))
else:
    freq_fit = freq_arr

x_centered = freq_fit - 3.0
poly_init = np.polyfit(x_centered, y_target, deg=4)[::-1]  # [a0, a1, a2, a3, a4] # Initial polynomial coefficients for the MS baseline, fitted to the target spectrum. The coefficients are ordered from constant term (a0) to quartic term (a4) and will serve as initial guesses for the combined model fitting.

# Initial guesses: [a0, a1, a2, a3, a4, p0, p1, p2, p3, p4, p5, p6, p7]
p0_combined = [
    poly_init[0], poly_init[1], poly_init[2], poly_init[3], poly_init[4],
    1.8217e-8,       # p0: Amp scale
    0.9398,          # p1: Numerator chirp frequency
    0.1358,          # p2: Frequency shift
    1.5594,          # p3: Exp scale
    0.4218,          # p4: Phase offset
    0.0852,          # p5: DC offset
    0.8743,          # p6: Denom scale
    0.1740           # p7: Denom constant offset
] # Initial guesses for the combined model parameters

# Physical bounds: broad on polynomial baseline, restricted on PySR modes
lower_bounds = [
    1.0,   -5.0,  -5.0,  -5.0,  -5.0,                       # a0 - a4
    0.0,    0.1,  -1.0,   0.1,  -np.pi, -1.0,  0.1,  -5.0   # p0 - p7
]

upper_bounds = [
    5.0,    5.0,   5.0,   5.0,   5.0,                       # a0 - a4
    1e-5,  10.0,   1.0,  10.0,   np.pi,  1.0, 10.0,   5.0   # p0 - p7
]

popt_ms_pysr, pcov_ms_pysr = curve_fit(
    ms_pysr_combined_model,
    freq_fit,
    y_target,
    p0=p0_combined,
    bounds=(lower_bounds, upper_bounds),
    method='trf',
    x_scale='jac',
    ftol=1e-15,
    xtol=1e-15,
    maxfev=250000
) # Fit the combined model (maximally smooth polynomial + 8-parameter PySR) to the target spectrum using non-linear least squares optimization. The initial parameter guesses, bounds, and fitting options are specified to ensure a robust fit while avoiding unphysical parameter values. The optimized parameters (popt_ms_pysr) and their covariance matrix (pcov_ms_pysr) are returned for further analysis and uncertainty estimation.

perr_ms_pysr = np.sqrt(np.diag(pcov_ms_pysr)) # Calculate the 1-sigma uncertainties of the fitted parameters from the diagonal elements of the covariance matrix. The square root of each diagonal element gives the standard deviation (1-sigma error) for the corresponding parameter, which provides an estimate of the uncertainty in the fitted values.

param_names = [
    'a0 (Offset)', 'a1 (Slope)', 'a2 (Quad)', 'a3 (Cubic)', 'a4 (Quartic)',
    'p0 (Amp Scale)', 'p1 (Chirp Freq)', 'p2 (Shift)', 'p3 (Exp Scale)',
    'p4 (Phase Offset)', 'p5 (DC Offset)', 'p6 (Denom Scale)', 'p7 (Denom Offset)'
]

print("="*68)
print(f"{'Parameter':<20} | {'Fitted Value':<18} | {'1-Sigma Error':<18}")
print("="*68)
for name, val, err in zip(param_names, popt_ms_pysr, perr_ms_pysr):
    print(f"{name:<20} | {val:<18.7e} | ± {err:<18.7e}")
print("="*68)

model_full = ms_pysr_combined_model(freq_fit, *popt_ms_pysr)
residuals = y_target - model_full
sigma_res = np.std(residuals)

fig, (ax1, ax2) = plt.subplots(
    2, 1, figsize=(10, 7), sharex=True, 
    gridspec_kw={'height_ratios': [3, 1], 'hspace': 0.08}
)

ax1.plot(freq_fit, y_target, 'k-', lw=2, alpha=0.6, label='Original Spectrum')
ax1.plot(freq_fit, model_full, '--', color='#D83A56', lw=2, label='MS Order 4 + 8p PySR Fit')
ax1.set_ylabel('Temperature [K]', fontsize=13, fontweight='bold')
ax1.set_title('Combined Model Fit: Order 4 MS Polynomial + 8-Parameter PySR', fontsize=15, fontweight='bold')
ax1.grid(True, linestyle='--', alpha=0.6)
ax1.legend(loc='upper right', fontsize=11)

# Normalized pulls
pulls = residuals / sigma_res
ax2.plot(freq_fit, pulls, '.', color='#2D3748', markersize=3, alpha=0.6)
ax2.axhline(0, color='#D83A56', ls='--', lw=1.2)
ax2.axhline(1, color='gray', ls=':', lw=0.9)
ax2.axhline(-1, color='gray', ls=':', lw=0.9)
ax2.set_ylabel(r'Pull [$\sigma$]', fontsize=11, fontweight='bold')
ax2.set_xlabel('Frequency [GHz]', fontsize=13, fontweight='bold')
ax2.set_ylim(-4, 4)
ax2.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

# MCMC for the 12 parameter equation (pysr+msfunction)

```python
import emcee
import corner
import numpy as np
import matplotlib.pyplot as plt
def ms_pysr_model_mcmc(nu, a0, a1, a2, a3, a4, log10_p0, p1, p2, p3, p4, p5, p6, p7):
    x = nu - 3.0
    t_ms = a0 + a1 * x + a2 * x**2 + a3 * x**3 + a4 * x**4
    
    p0 = 10.0**log10_p0
    exp_term = np.exp(p3 / np.log(nu))
    
    arg_num = p1 * (nu + p2) * (nu - exp_term) - p4
    numerator = -p0 * (np.cos(arg_num) + p5)
    
    arg_den = nu * (-nu + exp_term)
    denominator = p6 * nu + np.cos(nu * np.log(nu)) - p7 + (p6 * np.cos(arg_den)) / nu
    denom_clipped = np.where(np.abs(denominator) < 1e-12, np.sign(denominator) * 1e-12, denominator)
    
    return t_ms + (numerator / denom_clipped)

ndim = 13
param_labels = [
    r"$a_0$", r"$a_1$", r"$a_2$", r"$a_3$", r"$a_4$",
    r"$\log_{10}(p_0)$", r"$p_1$", r"$p_2$", r"$p_3$",
    r"$p_4$", r"$p_5$", r"$p_6$", r"$p_7$"
]

# Mode vector from curve_fit
theta_mode = np.array([
     2.7440171e+00, -9.7049314e-02,  4.2591263e-02, -2.7471270e-02,  1.2937726e-02,
    -9.572,          8.8358,         0.7611,         9.7264,        -2.5360,
     0.9656,         9.9559,        -4.9896
])
prior_sigmas = np.array([
    0.05, 0.01, 0.01, 0.01, 0.01,  # Polynomial a0-a4
    0.5,  0.5,  0.2,  0.5,  0.2,   # PySR parameters
    0.2,  0.5,  0.5
])

lower_bounds = np.array([
     2.0,  -1.0,  -0.5,  -0.5,  -0.5,
    -12.0,  5.0,  -0.5,   6.0,  -np.pi,
     0.0,   5.0, -10.0
])
upper_bounds = np.array([
     3.5,   1.0,   0.5,   0.5,   0.5,
    -7.0,  12.0,   2.0,  13.0,   0.0,
     2.0,  15.0,   0.0
])

def log_prior(theta):
    if not np.all((theta >= lower_bounds) & (theta <= upper_bounds)):
        return -np.inf
    diff = (theta[5:] - theta_mode[5:]) / prior_sigmas[5:]
    return -0.5 * np.sum(diff**2)

def log_likelihood(theta, x, y, yerr):
    model = ms_pysr_model_mcmc(x, *theta)
    if not np.all(np.isfinite(model)):
        return -np.inf
    inv_sigma2 = 1.0 / (yerr**2)
    return -0.5 * np.sum((y - model)**2 * inv_sigma2)

def log_probability(theta, x, y, yerr):
    lp = log_prior(theta)
    if not np.isfinite(lp):
        return -np.inf
    ll = log_likelihood(theta, x, y, yerr)
    return lp + ll if np.isfinite(ll) else -np.inf
y_target = np.asarray(spectrum_with_CRR if 'spectrum_with_CRR' in locals() else yres_fit)
freq_fit = np.linspace(2.0, 4.0, len(y_target))

model_init = ms_pysr_model_mcmc(freq_fit, *theta_mode)
residuals = y_target - model_init
sigma_noise = np.std(residuals)
yerr = np.full_like(y_target, sigma_noise)

print(f"Per-channel RMS residual (noise estimate): {sigma_noise:.4e} K")
nwalkers = 64
max_steps = 15000

np.random.seed(42)
pos = theta_mode + 1e-4 * np.abs(theta_mode) * np.random.randn(nwalkers, ndim)
for i in range(ndim):
    pos[:, i] = np.clip(pos[:, i], lower_bounds[i] + 1e-5, upper_bounds[i] - 1e-5)

sampler = emcee.EnsembleSampler(
    nwalkers,
    ndim,
    log_probability,
    args=(freq_fit, y_target, yerr)
)

print(f"Starting MCMC sampling with {nwalkers} walkers on {ndim} parameters...")
sampler.run_mcmc(pos, max_steps, progress=True)
print(f"Completed {max_steps} iterations.")
tau = sampler.get_autocorr_time(quiet=True)
valid_tau = np.all(np.isfinite(tau))

if not valid_tau or np.any(sampler.iteration < 50 * tau):
    print("The chain is shorter than 50 times the integrated autocorrelation time for 13 parameter(s). Use this estimate with caution and run a longer chain!")
    print(f"N/50 = {sampler.iteration // 50}; tau: {tau}")

burnin = int(2.0 * np.max(tau)) if valid_tau else 1336
thin = int(0.5 * np.min(tau)) if valid_tau else 226
thin = max(1, thin)

print("\n--- Diagnostics Summary ---")
print(f"Mean Acceptance Fraction : {np.mean(sampler.acceptance_fraction):.3f}")
for name, t in zip(param_labels, tau):
    print(f"Parameter {name:16s} : tau = {t:6.2f} steps")
print(f"Burn-in Discarded        : {burnin} steps")
print(f"Thinning Factor          : Every {thin} steps\n")

chain = sampler.get_chain()
fig, axes = plt.subplots(ndim, 1, figsize=(10, 1.4 * ndim), sharex=True)
for i in range(ndim):
    axes[i].plot(chain[:, :, i], color="k", alpha=0.2, lw=0.4)
    axes[i].axhline(theta_mode[i], color="#D83A56", linestyle="--", lw=1.0)
    axes[i].axvline(burnin, color="blue", linestyle=":", lw=1.0)
    axes[i].set_ylabel(param_labels[i], fontsize=10)
    axes[i].yaxis.set_label_coords(-0.08, 0.5)

axes[-1].set_xlabel("Iteration Step", fontsize=11)
plt.tight_layout()
plt.show()

flat_samples = sampler.get_chain(discard=burnin, thin=thin, flat=True)

samples_physical = np.copy(flat_samples)
samples_physical[:, 5] = 10.0**samples_physical[:, 5]  # Convert log10(p0) to physical p0

labels_physical = [
    r"$a_0$", r"$a_1$", r"$a_2$", r"$a_3$", r"$a_4$",
    r"$p_0$", r"$p_1$", r"$p_2$", r"$p_3$",
    r"$p_4$", r"$p_5$", r"$p_6$", r"$p_7$"
]

print("--- Physical Parameter Constraints (Median ± 1-sigma) ---")
for i in range(ndim):
    percentiles = np.percentile(samples_physical[:, i], [16, 50, 84])
    q = np.diff(percentiles)
    print(f"{labels_physical[i]:16s} : {percentiles[1]:14.7e} (+{q[1]:.7e}, -{q[0]:.7e})")

fig = corner.corner(
    flat_samples,
    labels=param_labels,
    truths=theta_mode,
    truth_color="#D83A56",
    quantiles=[0.16, 0.50, 0.84],
    show_titles=True,
    title_fmt=".3f",
    title_kwargs={"fontsize": 8},
    smooth=1.0,
    plot_datapoints=False,
    fill_contours=True,
    levels=(0.68, 0.95)
)
plt.show()
```
