```python
file_input_path = '/home/lenovo/NEw_SKY 1' # to read the input file
file_output_path = '/home/lenovo/NEw_SKY 1/result'# to save the output file
recomb_file = os.path.join(file_input_path,'total_spec_new1.dat') # to read the recombination line file
df = pd.read_csv(recomb_file,sep= "    " ,header = None) # to read the recombination line file
df.reset_index(drop=True, inplace=True) # to reset the index of the dataframe
df.columns = ["Frequency","Intensity"] # to rename the columns of the dataframe
x = df["Frequency"] # to extract the frequency column
y= df["Intensity"] # to extract the intensity column
ius = InterpolatedUnivariateSpline(x, y)   #1-D interpolating spline for a given set of data points; default k=3 being cubic spline
CRR_Int = ius(x) # to interpolate the intensity values at the frequency points
```

```python
from pysr import PySRRegressor # to import the PySRRegressor class from the pysr module
crr_data = np.loadtxt(os.path.join(file_input_path, "total_spec_new1.dat")) # to load the data from the file "total_spec_new1.dat" into a numpy array
print("Shape:", crr_data.shape) # to print the shape of the numpy array
print("First 3 rows:", crr_data[:3]) # to print the first 3 rows of the numpy array
```

```python
freq_GHz  = crr_data[:, 0]   # to extract the frequency column from the numpy array
intensity = crr_data[:, 1]   # to extract the intensity column from the numpy array
mask = (freq_GHz >= 2.0) & (freq_GHz <= 4.0) # to create a mask for the frequency range, so that only the frequency values between 2.0 GHz and 4.0 GHz are considered
freq_GHz  = freq_GHz[mask] # to filter the frequency values
intensity = intensity[mask] # to filter the intensity values
freq_norm = freq_GHz / 3.0   # to normalise the frequency values to be around 1.0 (3 GHz is the centre of the range)       
intens_norm = intensity / 1e-27     # to normalise the intensity values, where 1e-27 is a scaling factor to bring the values into a reasonable range for PySR

X = freq_norm.reshape(-1, 1)        # to reshape the normalised frequency values into a 2D array for PySR
y = intens_norm     # to use the normalised intensity values as the target variable for PySR

print(f"X range: {X.min():.3f} – {X.max():.3f}") # to print the range of the normalised frequency values
print(f"y range: {y.min():.4f} – {y.max():.4f}")# to print the range of the normalised intensity values

model = PySRRegressor(
    populations      = 100, # to set the number of models in each generation of the genetic algorithm
    niterations      = 300, # to set the number of generations for the genetic algorithm
    binary_operators = ["+", "-", "*"], # to set the binary operators for the symbolic regression
    unary_operators  = ["sin","cos"],# to set the unary operators for the symbolic regression
    maxsize          = 50, # to set the maximum size of the symbolic expressions
    parsimony=0.01,
    #complexity_of_constants=4,
    model_selection  = "best", # to set the model selection criterion to choose the best model based on validation loss
    nested_constraints = {
        "sin": {"sin": 1},
        "cos": {"cos":1},
    }, # to set the nested constraints for the symbolic regression to avoid nested sin and cos functions
    should_optimize_constants=True, # Ensures optimization loops pull constants toward exact physical values
    random_state=42,       # Sets the PRNG seed across Python and Julia, where PRNG = Pseudo-Random Number Generator
    deterministic=True,    # Instructs the backend to enforce reproducibility
    parallelism="serial",   # serial is more reproducible than multithreaded, but slower. Use "threaded" for speed.thread is a good compromise between speed and reproducibility. Use "async" for maximum speed, but results will be non-deterministic.
    verbosity        = 0, # to set the verbosity level of the output
    tempdir          = "/tmp/pysr", # to set the temporary directory for the symbolic regression
)
model.fit(X, y, variable_names=["f"])
```

```python
import matplotlib.pyplot as plt

y_pred_norm = model.predict(X)
y_pred_phys = y_pred_norm * 1e-27   # back to J/m²/s/Hz/Sr

plt.figure(figsize=(9, 5))
plt.plot(freq_GHz, intensity,      'r-',  lw=2,   label="CRR data (from notebook)")
plt.plot(freq_GHz, y_pred_phys,    'b--', lw=1.8, label=f"PySR fit: {model.sympy()}")
plt.xlabel("Frequency (GHz)", fontsize=13)
plt.ylabel("Intensity (J m⁻² s⁻¹ Hz⁻¹ Sr⁻¹)", fontsize=12)
plt.title("Recombination Intensity — PySR Symbolic Fit", fontsize=14)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
freq_GHz  = crr_data[:, 0]   # to extract the frequency column from the numpy array
intensity = crr_data[:, 1]   # to extract the intensity column from the numpy array
mask = (freq_GHz >= 2.0) & (freq_GHz <= 4.0) # to create a mask for the frequency range, so that only the frequency values between 2.0 GHz and 4.0 GHz are considered
freq_GHz  = freq_GHz[mask] # to filter the frequency values
intensity = intensity[mask] 
c = 2.99792458e8 # to set the speed of light in m/s
k = 1.3806488e-23 # to set the Boltzmann constant in J/K
freq_Hz  = freq_GHz * 1e9 # to convert the frequency values from GHz to Hz, we are converting intensity values from J/m²/s/Hz/Sr to K using the Rayleigh-Jeans law
temp_RJ  = intensity * c**2 / (2 * k * freq_Hz**2)   # in Kelvin (~1e-7 scale)
T_mean = temp_RJ.mean() # to calculate the mean of the temperature values
temp_norm = temp_RJ / T_mean       # to normalise the temperature values

X_T = freq_norm.reshape(-1, 1)     # to reshape the normalised frequency values into a 2D array for PySR
y_T = temp_norm# to use the normalised temperature values as the target variable for PySR

print(f"T_mean = {T_mean:.4e}") # to print the mean of the temperature values
print(f"X range: {X_T.min():.3f} – {X_T.max():.3f}") # to print the range of the normalised frequency values
print(f"y range: {y_T.min():.4f} – {y_T.max():.4f}") # to print the range of the normalised temperature values
model_T = PySRRegressor(
    populations      = 150,
    niterations      = 350,
    binary_operators = ["+", "-", "*","/"],
    unary_operators  = ["sin", "cos ","exp","log"],
    maxsize          = 50,
    model_selection  = "best",
    parsimony=0.01,
    complexity_of_constants=4,
    random_state     = 42,
    deterministic    = True,
    parallelism      = "serial",
    nested_constraints = {
        "sin": {"sin": 0},
        "cos": {"cos": 0},
        "exp": {"exp": 0},
        "log": {"log": 0}
    },
    verbosity        = 0,
    tempdir          = "/tmp/pysr",
)
model_T.fit(X_T, y_T, variable_names=["f"])

print(f"\nBest equation (normalised): T_norm(f) = {model_T.sympy()}")# to print the best equation for the normalised temperature values
print(f"Physical: T(freq_GHz) = ({model_T.sympy()}) × {T_mean:.4e}  K") # to print the best equation for the physical temperature values
print(f"where f = freq_GHz / 3.0")# to print the normalisation factor for the frequency values

from sklearn.metrics import r2_score
y_T_pred = model_T.predict(X_T)
print(f"\nR² = {r2_score(y_T, y_T_pred):.6f}")
```

```python
import numpy as np
from scipy.optimize import curve_fit
freq_GHz = crr_data[:, 0]  # assuming column 0 is frequency, in GHz already
intensity = crr_data[:, 1]

mask = (freq_GHz >= 2.0) & (freq_GHz <= 4.0)
x_phys= freq_GHz[mask] 
y_phys=intensity[mask]
X_data = freq_GHz[mask] / 3.0
y_target = intensity[mask] / 1e-27
lower_bounds = [0.0,  0.0,  0.0, -10.0,  10.0]
upper_bounds = [1.0,  0.1,  1.0,   0.0, 100.0]
def CRR_clean_fit(f, p1, p2, p3, A, B):
    return p2 * f * np.cos(A * f**3 + B * f) + p1 * f + p3 # to define the function for the curve fitting, where p1, p2, p3 are the independent constants, and A and B are the degenerate constants that are re-parameterized as A = p0 * p4 and B = p5 * p4.
# A = p0 * p4 = 0.779887 * (-8.065463) ≈ -6.289
# B = p5 * p4 = -5.399201 * (-8.065463) ≈ 43.547
# original values: p1 = 0.336907, p2 = 0.031574, p3 = 0.327928, A = -6.289, B = 43.547
p0_guess = [0.3, 0.03, 0.3, -6.2, 43] # this is to provide initial guesses for the parameters p1, p2, p3, A, and B based on previous calculations or estimates. These initial guesses help the curve fitting algorithm converge to a solution more efficiently.

popt, pcov = curve_fit(CRR_clean_fit, X_data, y_target, p0=p0_guess,bounds=(lower_bounds, upper_bounds))# to perform the curve fitting using the CRR_clean_fit function, the normalized frequency values (X_data), the normalized intensity values (y_data), and the initial guesses for the parameters (p0_guess). The function returns the optimal values for the parameters (popt) and the covariance of the parameters (pcov).
perr = np.sqrt(np.diag(pcov)) # to calculate the standard deviation (errors) of the fitted parameters from the diagonal of the covariance matrix (pcov). The square root of the diagonal elements gives the uncertainties associated with each fitted parameter.
param_names = ['p1', 'p2', 'p3', 'A', 'B']
for name, val, err in zip(param_names, popt, perr):
    print(f"{name}: {val:.6f} ± {err:.6f}")
```

```python
import numpy as np
import scipy.stats as stats
from sklearn.metrics import r2_score

y_pysr = CRR_clean_fit(X_data, *popt)   # use X_data and popt from cell above

y_err = np.ones_like(y_target) * np.std(y_target - y_pysr)  # residual-based

print(f"Estimated noise (y_err): {y_err[0]:.6f}")
print(f"As % of mean signal:     {100*y_err[0]/np.mean(y_target):.2f}%")

num_parameters     = 5
chi_squared        = np.sum(((y_target - y_pysr) / y_err) ** 2)
degrees_of_freedom = len(y_target) - num_parameters
reduced_chi2       = chi_squared / degrees_of_freedom
p_value            = 1 - stats.chi2.cdf(chi_squared, degrees_of_freedom)

print(f"\nChi-squared (χ²):          {chi_squared:.4f}")
print(f"Degrees of freedom (ν):    {degrees_of_freedom}")
print(f"Reduced chi-squared (χ²ᵥ): {reduced_chi2:.4f}")
print(f"p-value:                   {p_value:.6f}")
print(f"R²:                        {r2_score(y_target, y_pysr):.6f}")
```

```python
import numpy as np
import os
from scipy.optimize import curve_fit

mask = (freq_GHz >= 2.0) & (freq_GHz <= 4.0)
f_phys = freq_GHz[mask]
i_phys = intensity[mask]
c = 2.99792458e8
k = 1.3806488e-23
freq_Hz = f_phys * 1e9
temp_phys = i_phys * c**2 / (2 * k * freq_Hz**2)# to convert the intensity values to Rayleigh-Jeans temperature in Kelvin using the formula T = I * c² / (2 * k * f²), where I is the intensity, c is the speed of light, k is the Boltzmann constant, and f is the frequency in Hz.
f_norm = f_phys / 3.0
def crr_temperature_profile(f, scale, a, b, c, d): # TO define the function for the curve fitting, where scale, a, b, c, and d are the parameters to be fitted. The function represents the temperature profile as a function of frequency (f) based on the PySR symbolic regression output.
    # T_norm(f) = log(f + a * sin(b/f) * cos(f^2) + c * exp(-d*f))
    inner_term = f + a * np.sin(b / f) * np.cos(f**2) + c * np.exp(-d * f)
    return scale * np.log(np.clip(inner_term, 1e-12, None))
initial_guesses = [2.75e-7, 0.21, 21.06, 37.09, 3.29]# to provide initial guesses for the parameters scale, a, b, c, and d based on previous calculations or estimates. These initial guesses help the curve fitting algorithm converge to a solution more efficiently.
popt, pcov = curve_fit(
    crr_temperature_profile, f_norm, temp_phys, p0=initial_guesses# TO perform the curve fitting using the crr_temperature_profile function, the normalized frequency values (f_norm), the physical temperature values (temp_phys), and the initial guesses for the parameters (initial_guesses). The function returns the optimal values for the parameters (popt) and the covariance of the parameters (pcov)
)
perr = np.sqrt(np.diag(pcov)) # to calculate the standard deviation (errors) of the fitted parameters from the diagonal of the covariance matrix (pcov). The square root of the diagonal elements gives the uncertainties associated with each fitted parameter.
param_names = ["Scale", "a", "b", "c", "d"]
for name, val, err in zip(param_names, popt, perr):
    print(f"{name}: {val:.6e} ± {err:.6e}")
```

```python
y_pred = crr_temperature_profile(f_norm, *popt) # to calculate the predicted temperature values using the fitted parameters (popt) and the crr_temperature_profile function. The predicted values are stored in the variable y_pred.
y_err = np.std(temp_phys - y_pred)# to calculate the standard deviation of the residuals (temp_phys - y_pred) to estimate the noise in the data. The result is stored in the variable y_err.
num_params = len(popt)# to determine the number of parameters being fitted.
dof = len(temp_phys) - num_params# to calculate the degrees of freedom for the chi-squared test.
chi_squared = np.sum(((temp_phys - y_pred) / y_err) ** 2)# to calculate the chi-squared statistic by summing the squared normalized residuals (temp_phys - y_pred) / y_err. The result is stored in the variable chi_squared.
reduced_chi_squared = chi_squared / dof
r2 = r2_score(temp_phys, y_pred)
p_value = 1 - stats.chi2.cdf(chi_squared, dof)

print(f"Reduced Chi-squared (χ²ᵥ): {reduced_chi_squared:.4f}")
print(f"R² Value:                  {r2:.6f}")
print(f"Chi-squared (χ²):          {chi_squared:.4f}")
print(f"Degrees of Freedom (ν):    {dof}")
print(f"p-value:                   {p_value:.6f}")
```

```python
import numpy as np
import emcee
import corner
import matplotlib.pyplot as plt

def model_func(f, scale, a, b, c, d):
    inner = f + a * np.sin(b / f) * np.cos(f**2) + c * np.exp(-d * f)
    return scale * np.log(np.clip(inner, 1e-12, None))# PYSR expression for the temperature profile, where f is the frequency and scale, a, b, c, d are the parameters to be fitted. The function returns None if any value of the inner term is less than or equal to zero to avoid taking the logarithm of non-positive values.

def log_prior(theta):
    scale, a, b, c, d, log_sigma = theta
    # Hard bounds based on curve_fit boundaries and physical range
    if (0 < scale < 1e-5 and 
        0 < a < 1.0 and 
        10.0 < b < 30.0 and 
        0.0 < c < 100.0 and 
        0.0 < d < 10.0 and 
        -25.0 < log_sigma < -10.0): # where log_sigma is the logarithm of the standard deviation of the noise in the data. The bounds for log_sigma are set to ensure that the noise level is within a reasonable range for the MCMC sampling.
        return 0.0
    return -np.inf# prior function that returns 0.0 if the parameters are within the specified bounds, and -np.inf (negative infinity) if any parameter is outside the bounds. This effectively imposes hard constraints on the parameter space for the MCMC sampling.

def log_likelihood(theta, f, y):
    scale, a, b, c, d, log_sigma = theta
    model = model_func(f, scale, a, b, c, d)
    sigma2 = np.exp(2.0 * log_sigma)
    # Gaussian log-likelihood
    return -0.5 * np.sum(((y - model)**2) / sigma2 + 2.0 * log_sigma + np.log(2.0 * np.pi))# to calculate the log-likelihood of the model given the parameters theta, frequency f, and observed data y. The function computes the model predictions using model_func, checks for validity, and then calculates the log-likelihood based on a Gaussian noise model with variance sigma2.

def log_posterior(theta, f, y):
    lp = log_prior(theta)
    if not np.isfinite(lp):
        return -np.inf
    return lp + log_likelihood(theta, f, y)# to calculate the log-posterior probability of the parameters theta given the frequency f and observed data y. The function first computes the log-prior (lp) using log_prior, and if lp is finite, it adds the log-likelihood (computed using log_likelihood) to return the total log-posterior.

labels = ["scale", "a", "b", "c", "d", "log_sigma"]
ndim = len(labels)
nwalkers = 32# to set the number of walkers for the MCMC sampling. Walkers are individual chains that explore the parameter space in parallel, and having multiple walkers helps to ensure better exploration and convergence of the MCMC algorithm.
nsteps = 4000# to set the number of steps for each walker in the MCMC sampling. Each walker will take nsteps steps in the parameter space, and the total number of samples collected will be nwalkers * nsteps.
burnin = 1000# to set the number of initial steps to discard as burn-in. The burn-in period allows the walkers to converge to the target distribution before collecting samples for analysis. The first burnin steps of each walker will be discarded, and only the remaining samples will be used for posterior analysis.

sigma_guess = np.std(temp_RJ - crr_temperature_profile(f_norm, *popt))# to calculate an initial guess for the standard deviation of the noise in the data. The guess is based on the standard deviation of the residuals between the observed temperature data (temp_RJ) and the model predictions using the fitted parameters (popt) from the curve fitting.
theta_initial = np.append(popt, np.log(sigma_guess))# to create an initial parameter vector (theta_initial) for the MCMC sampling. The vector consists of the fitted parameters (popt) from the curve fitting, followed by the logarithm of the estimated noise standard deviation (log(sigma_guess)). This initial guess will be used to initialize the walkers in the MCMC algorithm.

pos = theta_initial + 1e-4 * np.abs(theta_initial) * np.random.randn(nwalkers, ndim)# to initialize the positions of the walkers in the MCMC sampling. Each walker's initial position is set to be a small random perturbation (1e-4 times the absolute value of theta_initial) around the initial parameter vector (theta_initial). The random perturbations are drawn from a standard normal distribution (np.random.randn) to ensure that the walkers start exploring the parameter space from slightly different locations.

sampler = emcee.EnsembleSampler(nwalkers, ndim, log_posterior, args=(f_norm, temp_RJ))# to create an instance of the emcee.EnsembleSampler class, which will be used to perform the MCMC sampling. The sampler is initialized with the number of walkers (nwalkers), the number of dimensions (ndim), the log-posterior function (log_posterior), and the arguments (f_norm, temp_RJ) that will be passed to the log-posterior function during sampling.
print("Sampling posterior space")
sampler.run_mcmc(pos, nsteps, progress=True)# to run the MCMC sampling using the sampler instance. The walkers will start from their initial positions (pos) and take nsteps steps in the parameter space, exploring the posterior distribution defined by the log-posterior function. The progress of the sampling will be displayed in the console.

tau = sampler.get_autocorr_time(quiet=True)# to calculate the mean autocorrelation time (tau) of the MCMC chains. The autocorrelation time is a measure of how many steps it takes for the samples to become effectively independent. A smaller tau indicates that the samples are more independent, while a larger tau suggests that the samples are more correlated. The quiet=True argument suppresses warnings if the autocorrelation time cannot be estimated reliably.
print(f"Mean Autocorrelation Time (tau): {np.nanmean(tau):.1f} steps")

flat_samples = sampler.get_chain(discard=burnin, thin=15, flat=True)# to extract the flattened samples from the MCMC chains after discarding the burn-in steps and thinning the chains. The get_chain method retrieves the samples from the sampler, and the discard=burnin argument specifies how many initial steps to discard. The thin=15 argument indicates that every 15th sample will be kept to reduce autocorrelation, and flat=True flattens the resulting array into a 2D array where each row corresponds to a sample and each column corresponds to a parameter.

print("\nMCMC Parameter Constraints:")
for i in range(ndim):
    mcmc = np.percentile(flat_samples[:, i], [16, 50, 84])
    q = np.diff(mcmc)
    print(f"{labels[i]:>10s} = {mcmc[1]:.6e} (+{q[1]:.6e}, -{q[0]:.6e})")

fig = corner.corner(
    flat_samples, 
    labels=labels, 
    truths=theta_initial, 
    quantiles=[0.16, 0.5, 0.84], 
    show_titles=True,
    title_fmt=".3e",         # <-- formats titles as e.g. 2.862e-07
    title_kwargs={"fontsize": 10}
)
plt.show()
```
